```python
class Car: #class
    def __init__(self, brand, color): #constructor
        self.brand = brand
        self.color = color

    def display(self):                               #method
        print(f"{self.brand} car is {self.color}")

car1 = Car("Tesla","Black") #object
car1.display()
```

```python
#Encapsulation  means bundling data and methods together and restricting direct access to some data.
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)

print(acc.get_balance())
```

```python
#Inheritance allows one class to acquire the properties and methods of another class.
class Animal:
    def speak(self):
        print("Animal makes sound")

class Dog(Animal): # passing in the parent class
    pass

d = Dog()
d.speak()
```

```python
#Polymorphism means one interface, many forms.
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

class Cat(Animal):
    def sound(self):
        print("Meow")

Dog().sound()
Cat().sound()
#The same method sound() behaves differently for different objects.
```

```python
#An abstract class is a class that cannot be instantiated and may contain abstract methods. Abstract methods act like contracts, forcing child classes to provide their own implementation.
from abc import ABC, abstractmethod

# Abstract class (blueprint)
class Vehicle(ABC):

    # Must be implemented by child classes
    @abstractmethod
    def start(self):
        pass


# Child class
class Car(Vehicle):

    # Implementation of abstract method
    def start(self):
        print("Car started")


# Create object
c = Car()

# Call method
c.start()
```

Types of classes


```python
#1. Concrete Class - A normal class that can be instantiated.
class Car:
    def start(self):
        print("Car started")

c = Car()
```

```python
#2. Abstract Class - A class containing one or more abstract methods. Cannot be instantiated directly.
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

```

```python
#3. Base Class (Parent Class / Super Class) A class that is inherited by another class.
class Animal:
    pass
```

```python
#4. Derived Class (Child Class / Sub Class) - A class that inherits from another class.
class Animal:
    pass

class Dog(Animal):
    pass
```

```python
#5. Inner Class (Nested Class) - A class defined inside another class.
class Outer:

    class Inner:
        def show(self):
            print("Inside Inner")

obj = Outer.Inner()
obj.show()
```

```python
#6. Utility Class - A class containing only static methods and no object-specific data.
class MathUtil:

    @staticmethod
    def add(a, b):
        return a + b

print(MathUtil.add(2, 3))
```

```python
#7. Data Class - Used mainly to store data. Introduced with the dataclasses module.
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int

s = Student("Nalin", 22)
print(s)
```

```python
#An Enum (Enumeration) is used to define a fixed set of named constants. It makes code more readable and prevents using arbitrary values.
from enum import Enum

class Color(Enum):
    RED = 1
    GREEN = 2
    BLUE = 3

print(Color.RED)
print(Color.RED.name)
print(Color.RED.value)
```

```python
#A frozen dataclass is a dataclass whose objects become immutable after creation.
from dataclasses import dataclass

@dataclass(frozen=True)
class Employee:
    id: int
    name: str

emp = Employee(101, "Nalin")

print(emp.name)
```

```python
#A Factory Class is a class that encapsulates object creation logic and returns objects based on input, hiding the instantiation details from the client code.
class Dog:
    def speak(self):
        print("Bark")

class Cat:
    def speak(self):
        print("Meow")


# Factory class
class AnimalFactory:

    @staticmethod
    def get_animal(animal_type):

        if animal_type == "dog":
            return Dog()

        elif animal_type == "cat":
            return Cat()

        else:
            return None


animal = AnimalFactory.get_animal("dog")
animal.speak()
```

```python
class Database:

    def connect(self):
        print("Connected to database")


class DatabaseProxy:

    def __init__(self, user):
        self.user = user
        self.database = Database()

    def connect(self):

        if self.user == "admin":
            self.database.connect()
        else:
            print("Access Denied")

db = DatabaseProxy("admin")
db.connect()
```

```python
#A metaclass is a class whose instances are classes. It controls how classes are created, modified, and initialized. The default metaclass in Python is type.

class MyMeta(type):

    def __new__(cls, name, bases, attrs):
        print(f"Creating class: {name}")
        return super().__new__(cls, name, bases, attrs)

class Employee(metaclass=MyMeta):
    pass
```
