```python
#Matplotlib is the most popular Python library for creating charts, graphs, and visualizations.

import matplotlib.pyplot as plt
```

```python
#Line plot
x= [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(
    x,
    y,
    color='red',
    linestyle='--',
    marker='o',
    linewidth=3
)
plt.title("Sales Growth")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
#scatter plot

age = [18, 20, 22, 24, 26]
salary = [20000, 25000, 28000, 35000, 40000]

plt.scatter(age, salary)

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")

plt.show()
```

```python
#bar chart
products = ["Laptop", "Phone", "Tablet"]
sales = [150, 250, 100]

plt.bar(products, sales)
plt.show()
```

```python
#histogram - shows data distribution
marks = [70, 75, 80, 85, 90, 90, 75, 68, 72, 88]

plt.hist(marks, bins=5)

plt.title("Marks Distribution")
plt.show()
```

```python
#pie chart - used to show percentages

sizes = [30, 45, 25]
labels = ["Python", "Java", "C++"]

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%'
)

plt.show()
```

```python
plt.plot()      # Line chart
plt.scatter()   # Scatter plot
plt.bar()       # Bar chart
plt.hist()      # Histogram
plt.pie()       # Pie chart
plt.title()     # Title
plt.xlabel()    # X label
plt.ylabel()    # Y label
plt.legend()    # Legend
plt.grid()      # Grid
plt.savefig()   # Save image
plt.show()      # Display graph
```

```python
tips = sns.load_dataset("tips")

sns.scatterplot(data=tips,x="total_bill",y="tip")

plt.show()
```

```python
#box plot
'''
Shows:
Median
Minimum
Maximum
Outliers'''
sns.boxplot(
    x="day",
    y="total_bill",
    data=tips
)

plt.show()
```

```python
'''
Count Plot
Counts category occurrences.
'''
sns.countplot(
    x="day",
    data=tips
)

plt.show()
```
