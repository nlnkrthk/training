```python
#Missing values are usually represented as: Nan , None
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Marks": [90, np.nan, 80]
})

print(df)
```

```python
#Check Missing Values
print(df.isnull())
```

```python
#Count Missing Values
print(df.isnull().sum())
```

```python
#MultiIndex allows multiple levels of indexing.
#set_index() - Converts one or more columns into indexes.

df = pd.DataFrame({
    "Department":["IT","IT","HR"],
    "Employee":["A","B","C"],
    "Salary":[50000,60000,45000]
})

df = df.set_index(["Department","Employee"])

print(df)
```

```python
print(df.loc[("IT","A")])  #acessing data
```

```python
#reset_index() - Converts index back to normal columns.
df = df.reset_index()
```

```python
#Vectorization -  means performing operations on an entire column at once without loops.
df["Salary"] = df["Salary"] + 1000
print(df)
```

```python
# map() - used mainly with series to transform values
df = pd.DataFrame({
    "Gender":["M","F","M"]
})
df["Gender"] = df["Gender"].map({
    "M":"Male",
    "F":"Female"
})
print(df)
```

```python
df["Value"] = pd.Series([1,2,3])

df["Square"] = df["Value"].map(lambda x: x*x)

print(df)
```

```python
#Rolling Window (rolling()) - A rolling window looks at a fixed number of previous rows.
sales = pd.DataFrame({
    "Sales":[100,200,300,400,500]
})

sales["Moving Average"] = sales["Sales"].rolling(window=3).mean()

print(sales)
```

```python
'''Window functions perform calculations over a group of rows.
rolling()
expanding()
cumsum()
cummax()
cummin()
cumprod()
'''
```

```python
# expanding() - Uses all rows from the beginning up to the current row.

sales = pd.DataFrame({
    "Sales":[100,200,300,400,500]
})

sales["Running Average"] = sales["Sales"].expanding().mean()

print(sales)
```
