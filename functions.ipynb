```python
#Function with Default Parameters
def calculate_age (birth_year,current_year = 2021):
    age = current_year - birth_year
    return age
print('Age: ', calculate_age(1821))
```

```python
'''Arbitrary Number of Arguments
If we do not know the number of arguments we pass to our function, we can create a function which can take arbitrary number of arguments by adding * before the parameter name.'''
def sum_all_nums(*nums):
    total = 0
    for num in nums:
        total += num     # same as total = total + num
    return total
print(sum_all_nums(2, 3, 5)) # 10
```

```python
#**kwargs collects any number of keyword arguments into a dictionary.
def show_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_details(name="Nalin", age=22, city="Chennai")
```

```python
def posi(a, b, /, x, y):
    # a and b -> positional-only
    # x and y -> can be positional or keyword
    print(a, b, x, y)

# ✅ Valid
posi(1, 2, 3, 4)

# ✅ Valid
posi(1, 2, x=3, y=4)

# ❌ Invalid
# posi(a=1, b=2, x=3, y=4)
# Error: a and b cannot be passed as keywords
```

```python
def keyw(name, *, city):
    # name -> positional or keyword
    # city -> keyword-only
    print(f"{name} lives in {city}")

# ✅ Valid
keyw("Nalin", city="Hosur")

# ✅ Also valid
keyw(name="Nalin", city="Hosur")

# ❌ Invalid
# keyw("Nalin", "Hosur")
# Error: city must be passed as a keyword
```

```python
'''A lambda function is a short, one-line function without a name.'''
# Normal function
def square(x):
    return x * x

# Same using lambda
square = lambda x: x * x

print(square(5))  # 25
```

```python
add = lambda a, b: a + b

print(add(2, 3))  # 5
```

```python
def outer():
    def inner():
        print("I am in")
    return inner  # return the function itself

x = outer()  # x now stores the inner function
x()          # calls inner()
```

```python
def square(x):
    return x * x

def calc(fun, val):
    return fun(val)   # use val

print(calc(square, 5))
```

```python
#closure remembrance a function that remembers variables from its outer function even after the outer function has finished.
def outer(msg):
    def inner():
        print(msg)  # remembers msg
    return inner

hello = outer("Hello")
bye = outer("Bye")

hello()  # contains inner and now called
bye()    # Bye
```

```python
#Generator = memory-efficient way to produce values one at a time using yield.
def numbers():
    for i in range(1, 100000):
        yield i   # produce one value at a time

g = numbers()

print(next(g))  # 1
print(next(g))  # 2
print(next(g))  # 3
```

```python
def numbers():
    yield 1*2
    yield 2*3
g = numbers()
print(next(g))
print(next(g))
print(next(g,"none")) # returns default val
```

```python
#decorator. It adds extra behavior before and after a function runs.

def decorator(func):
    def wrapper():
        print("before")
        func()
        print("after")
    return wrapper

def display():
    print("welcome")

display = decorator(display)  # decorate display
display()
```

```python
def decorator(func):
    def wrapper():
        print("before")
        func()
        print("after")
    return wrapper

@decorator
def display():
    print("welcome")

display()
```
