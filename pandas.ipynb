```python
import pandas as pd
'''Pandas is a powerful Python library used for data manipulation, analysis, and processing. It helps you work with structured data such as tables, spreadsheets, CSV files, and databases.
Series - A one-dimensional labeled array that can hold any data type (integers, strings, floats, etc.). It is similar to a column in a spreadsheet or a database table.
DataFrame - A two-dimensional labeled data structure that can hold multiple Series (columns) of different data types. It is similar to a table in a relational database or an Excel spreadsheet.'''
```

```python
s1 = pd.Series(["nalin","jermone","shek"])
s2 = pd.Series([11,21,31])
print(s1)
print(s2)
```

```python
m1 = pd.Series([1,2,3,4,5],["one","two","three","four","five"]) # values and the keys
m1["one"] = 10
print(m1)
```

```python
m1["six"] = 6 #add new key and value
print(m1)
```

```python
new  =  pd.Series([7,8],["seven","eight"])
m1 = pd.concat([m1,new]) # add new series to the existing series
print(m1)
```

loc → Selects data using labels (row/column names).

iloc → Selects data using integer positions (indexes).

```python
n1 = pd.Series([1,2,3,4,5],index=["one","two","three","four","five"])

#df.iloc[row_position, column_position]
print(n1.iloc[0]) # access the first element of the series
print(n1.iloc[[0, 2]])

print(n1.loc["one":"four"]) # access the element with the key "one"
```

```python
print(n1.loc[n1>2]) # access the elements greater than 2
```

```python
n1.loc["six"] = 6 # add new key and value
print(n1)
```

```python
n1 = n1 + 10
print(n1) # add 10 to all the values in the series
```

```python
n1["six"] += 10 # add 10 to the value of the key "six"
print(n1)
```

```python
n1 = n1.drop("six") # remove the key and value
print(n1)
```

```python
print("maximum",n1.max()) # maximum value in the series
print("minimum",n1.min()) # minimum value in the series
print("sum",n1.sum()) # sum of all values in the series
print("mean",n1.mean()) # mean of all values in the series
print("median",n1.median()) # median of all values in the series
print("standard deviation",n1.std()) # standard deviation of all values in the series
print(n1.sort_index()) # sort the series by index
print(n1.sort_values()) # sort the series by values
```

```python
print(n1.values) # get the values of the series
print(n1.index) # get the index of the series
print(n1.shape) # get the shape of the series
print(n1.size) # get the size of the series
print(n1.ndim) # get the number of dimensions of the series
print(n1.head()) # get the first 5 elements of the series
print(n1.tail()) # get the last 5 elements of the series
```

```python
print(n1.describe()) # get the summary statistics of the series
print(n1.unique()) # get the unique values of the series
print(n1.nunique()) # get the number of unique values of the series
```

```python
#pivot_table() function is used to create a pivot table in pandas. A pivot table is a data summarization tool that is used to summarize and aggregate data in a tabular format. It allows you to group data by one or more columns and perform calculations on the grouped data.
#cross_tab() function is used to create a cross tabulation in pandas. A cross tabulation is a table that shows the relationship between two or more categorical variables. It allows you to see how the values of one variable are distributed across the values of another variable.
#melt() function is used to reshape a dataframe from wide format to long format. It allows you to convert columns into rows, making it easier to analyze and visualize the data.
#stack() function is used to reshape a dataframe from wide format to long format. It allows you to convert columns into rows, making it easier to analyze and visualize the data.
#unstack() function is used to reshape a dataframe from long format to wide format. It allows you to convert rows into columns, making it easier to analyze and visualize the data.
#cross_tab() function is used to create a cross tabulation in pandas. A cross tabulation is a table that shows the relationship between two or more categorical variables. It allows you to see how the values of one variable are distributed across the values of another variable.
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['A', 'A', 'B', 'B'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 20000, 60000, 25000]
})

print(df)
```

```python
#pivot table groups data and performs aggregation (sum, mean, count, etc.).
pd.pivot_table(
    df,
    values='Sales',
    index='Name',
    columns='Product',
    aggfunc='sum'
)
```

```python
#crosstab() Creates a frequency table showing relationships between categorical variables.

df = pd.DataFrame({
    'Gender': ['M','F','M','F','M'],
    'Result': ['Pass','Pass','Fail','Pass','Fail']
})

pd.crosstab(df['Gender'], df['Result'])
```

```python
# melt() Converts wide format → long format.
df = pd.DataFrame({
    'Name': ['A', 'B'],
    'Math': [90, 80],
    'Science': [85, 75]
})

print(df)
```

```python
pd.melt(
    df
)
```

```python
#stack() - Converts columns into rows by creating a hierarchical index.
df = pd.DataFrame({
    'Math': [90, 80],
    'Science': [85, 75]
}, index=['A', 'B'])

print(df)
```

```python
df.stack()
```

```python
#unstack() - reverses the stack operation, converting rows back into columns. It allows you to reshape a dataframe from long format to wide format, making it easier to analyze and visualize the data.
s = df.stack()

print(s)
```

```python
s.unstack()
```
