```python
#PUT method - replace/update and entire row
from fastapi import FastAPI

# Create a FastAPI application
app = FastAPI()

# This dictionary acts like a temporary database.
# The key (1, 2) is the user ID.
# Each user has a name and age.
users = {
    1: {"name": "John", "age": 20},
    2: {"name": "Alice", "age": 25}
}

# GET request to display all users
# URL: http://127.0.0.1:8000/users
@app.get("/users")
def get_users():
    # Return the entire users dictionary
    return users


# PUT request to update an existing user
# {user_id} is a path parameter
# Example: /users/1
@app.put("/users/{user_id}")
def update_user(user_id: int, name: str, age: int):

    # Check whether the given user ID exists
    if user_id not in users:
        return {"message": "User not found"}

    # Update the user's name
    users[user_id]["name"] = name

    # Update the user's age
    users[user_id]["age"] = age

    # Return a success message along with the updated user
    return {
        "message": "User updated successfully",
        "user": users[user_id]
    }
```

```python
#patch - update only what you want , name/age in one row

from fastapi import FastAPI

app = FastAPI()

# Temporary database
users = {
    1: {"name": "John", "age": 20},
    2: {"name": "Alice", "age": 25}
}

# Display all users
@app.get("/users")
def get_users():
    return users

# Update only the name
@app.patch("/users/{user_id}/name")
def update_name(user_id: int, name: str):

    # Check if user exists
    if user_id not in users:
        return {"message": "User not found"}

    # Update only the name
    users[user_id]["name"] = name

    return {
        "message": "Name updated successfully",
        "user": users[user_id]
    }

'''
id - 1
name - nalin

{
  "1": {
    "name": "nalin",
    "age": 20
  },
  "2": {
    "name": "Alice",
    "age": 25
  }
}'''
```

```python
#DELETE request is used to remove existing data.
from fastapi import FastAPI

app = FastAPI()

# Temporary database
users = {
    1: {"name": "John", "age": 20},
    2: {"name": "Alice", "age": 25}
}

# Display all users
@app.get("/users")
def get_users():
    return users

# Delete a user
@app.delete("/users/{user_id}")
def delete_user(user_id: int):

    # Check if the user exists
    if user_id not in users:
        return {"message": "User not found"}

    # Delete the user from the dictionary
    del users[user_id]

    # Return a success message
    return {
        "message": "User deleted successfully"
    }

'''
before
users = {
    1: {"name": "John", "age": 20},
    2: {"name": "Alice", "age": 25}
}

delete - 1

after
{
  "2": {
    "name": "Alice",
    "age": 25
  }
}
'''
```

```python
#simple crud operations
from fastapi import FastAPI

app = FastAPI()

users = {
    1: {"name": "John", "age": 20},
    2: {"name": "Alice", "age": 25}
}

# CREATE
@app.post("/users/{user_id}")
def create_user(user_id: int, name: str, age: int):
    users[user_id] = {"name": name, "age": age}
    return {"message": "User created"}

# READ
@app.get("/users")
def get_users():
    return users

# UPDATE
@app.put("/users/{user_id}")
def update_user(user_id: int, name: str, age: int):
    if user_id not in users:
        return {"message": "User not found"}

    users[user_id]["name"] = name
    users[user_id]["age"] = age
    return {"message": "User updated"}

# DELETE
@app.delete("/users/{user_id}")
def delete_user(user_id: int):
    if user_id not in users:
        return {"message": "User not found"}

    del users[user_id]
    return {"message": "User deleted"}
```

```python
#Pydantic is used to validate and organize the data that the client sends to your API.

# Import FastAPI to create our API application
from fastapi import FastAPI

# Import BaseModel from Pydantic to create data models
from pydantic import BaseModel


# Create a FastAPI application object
app = FastAPI()


# Create a Pydantic model (data structure) for a User
# This defines what data the API expects from the client
class User(BaseModel):

    # User's name should be a string
    name: str

    # User's age should be an integer
    age: int


# Create a POST API endpoint
# URL: /users
# This endpoint is used to create a new user
@app.post("/users")
def create_user(user: User):

    # The 'user' parameter receives JSON data from the request body
    # FastAPI automatically:
    # 1. Reads the JSON data
    # 2. Checks if the fields are correct
    # 3. Converts it into a User object


    # Return a JSON response
    return {
        "message": "User created",

        # Return the user data received from the client
        "user": user
    }
```

```python
#Async allows your program to start a task, wait for something slow to finish, and meanwhile do other work instead of sitting idle.
#await means: "Wait for this task to finish, but allow other tasks to run while waiting."

from fastapi import FastAPI
import asyncio

app = FastAPI()


# A simple async function
async def get_message():
    # Simulating a slow task (like database/API call)
    await asyncio.sleep(3)

    return "Data received after 3 seconds"


# Async API endpoint
@app.get("/message")
async def message():

    # Wait for the async function to complete
    result = await get_message()

    return {
        "message": result
    }
```

```python
#Nested datatypes in Pydantic mean that one model can contain another model.
from pydantic import BaseModel

class Address(BaseModel):
    city: str
    state: str

class Student(BaseModel):
    name: str
    address: Address
```

```python
#Generic[T] type - tells Python that this class is generic, meaning it accepts a type parameter.
from pydantic import BaseModel
from typing import Generic, TypeVar

T = TypeVar("T")

class Response(BaseModel, Generic[T]):
    success: bool
    data: T
```

```python
# field validator is a function in Pydantic that checks or modifies a field's value before the model is created.
from pydantic import BaseModel, field_validator

class Student(BaseModel):
    age: int

    @field_validator("age")
    @classmethod
    def validate_age(cls, value):
        if value < 18:
            raise ValueError("Age must be 18 or above")
        return value
```

```python
#error/exception handling in api execution
from fastapi import FastAPI, HTTPException

app = FastAPI()

students = {
    1: "John",
    2: "Alice",
    3: "David"
}

@app.get("/students/{student_id}")
def get_student(student_id: int):

    if student_id not in students:
        raise HTTPException(
            status_code=404,
            detail="Student Not Found"
        )

    return {
        "id": student_id,
        "name": students[student_id]
    }
```
