Machine Learning (ML) is a branch of Artificial Intelligence (AI) that enables computers to learn from data and make predictions or decisions without being explicitly programmed for every task.

Instead of writing fixed rules, we provide the computer with data, and it learns patterns that it can use to make predictions on new, unseen data.

1. Supervised Learning

Definition:
Supervised learning is a type of machine learning in which the model is trained using labeled data. Each training example includes both the input (features) and the correct output (label).

Types of supervised learning

Regression: Predicts continuous values.

Examples: House price prediction, temperature forecasting.

Classification: Predicts categories or classes.

Examples: Spam detection, disease diagnosis.

1. Multiclass Classification

Definition:
Multiclass classification is a classification problem where each data point belongs to exactly one class out of more than two possible classes.

One input → One label

Number of classes > 2

2. Multilabel Classification

Definition:
Multilabel classification is a problem where one data point can belong to multiple classes simultaneously.

One input → Multiple labels

3. Imbalanced Classification

Definition:
Imbalanced classification occurs when some classes have many more examples than others.

Example

Fraud detection

Class	Number of Transactions
Genuine	99,000
Fraud	1,000

Here,

Genuine = 99%
Fraud = 1%

The classes are imbalanced.

1 Linear Regression  
Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value by modeling the relationship between an input variable (feature) and an output variable (target) with a straight line.  
-The algorithm chooses the line that minimizes the sum of squared errors (least squares) between the actual values and the predicted values.  

y=wx+b  
 
y = Predicted output (target/dependent variable)  
x = Input feature (independent variable)  
w = Weight (slope of the line)  
b = Bias (intercept)  


2 - Multiple Linear Regression  

Multiple Linear Regression (MLR) is an extension of Simple Linear Regression where the model uses two or more input features to predict a continuous numerical value.

Predict salary using experience + education + skills + location

y=w1​x1​+w2​x2​+w3​x3​+b

y = Predicted output  
x1,x2,x3 = input features  
w1,w2,w3 = weights for each feature  
b = Bias (intercept)

3 - Polynomial Linear Regression  
 is an extension of Linear Regression that can model non-linear relationships between the input variable(s) and the target variable. Despite its name, it is still considered a linear model because it is linear in its coefficients (weights).

 degree 2
 y = ax2 + bx + c

 degree 3
 y = ax3 + bx2 + cx + d

Even though Polynomial Regression includes terms like x2 ,x3
, and higher powers, it is still called Linear Regression because the model is linear with respect to its coefficients (parameters).
a, b, and c are the coefficients (parameters) that the model learns.
These coefficients are added together and multiplied by the input features, but they are never multiplied by each other or raised to powers.
 

Why do we need Regularization?

Suppose we have a Linear Regression model:

y=mx+c

or with multiple features:

If the model becomes too complex, it may memorize the training data instead of learning the underlying pattern. This is called overfitting.

Regularization solves this problem by penalizing large coefficient values, encouraging the model to keep the coefficients smaller.

4 - Ridge Regression (L2 Regularization)

Definition:
Ridge Regression is a regularized version of Linear Regression that adds an L2 penalty (sum of squared coefficients) to the cost function. It reduces the magnitude of the coefficients to prevent overfitting but does not eliminate any features.

The cost function becomes:

Cost=MSE+λ∑w

where

MSE = prediction error  
wi = model coefficients  
λ = regularization parameter

5 - Lasso Regression (L1 Regularization)

Definition:  
Lasso Regression is a regularized version of Linear Regression that adds an L1 penalty (sum of the absolute values of the coefficients) to the cost function. It shrinks coefficient values and can reduce some coefficients exactly to zero, effectively performing feature selection.

Equation
Cost=MSE+λ∑∣wi∣
Notice the absolute value instead of the square.



6 - Elastic Net Regression

Definition:
Elastic Net Regression is a regularized version of Linear Regression that combines both L1 (Lasso) and L2 (Ridge) penalties. It simultaneously shrinks coefficient values and performs feature selection, making it effective when dealing with many or highly correlated features.



7 - Decision Tree Regression
Definition

Decision Tree Regression is a supervised machine learning algorithm used to predict continuous numerical values by recursively splitting the dataset into smaller regions based on feature values. The prediction for each region is the average (mean) of the target values in that region.

Each question splits the data into smaller groups until similar target values are grouped together.

8 - Random Forest Regression
Definition

Random Forest Regression is a supervised machine learning algorithm that predicts continuous numerical values by combining the predictions of multiple Decision Trees. Each tree is trained on a random subset of the data and features, and the final prediction is the average of all the trees' predictions.

 9 - Support Vector Regression (SVR)
Definition

Support Vector Regression (SVR) is a supervised machine learning algorithm used to predict continuous numerical values. Instead of minimizing the error for every training point, SVR tries to find the best-fit line (or curve) that keeps most data points within a specified margin of error (ε) while maximizing the margin around the prediction function.
