Exception
An exception is a runtime problem that can be anticipated and handled using try-except.

Error
An error generally refers to serious issues that prevent the program from executing correctly. These are often caused by mistakes in the code itself.

```python
#Exception handling is a way to manage runtime errors in a program so that it doesn't crash unexpectedly. Python uses the try, except, else, and finally blocks for exception handling.
```

```python
try:
    # Code that may cause an exception
    result = 10 / 0

except ZeroDivisionError:
    # Code to handle the exception
    print("Cannot divide by zero")
```

```python
#Handling Multiple Exceptions
try:
    num = int(input("Enter a number: "))
    result = 100 / num

except ValueError:
    print("Invalid input. Please enter a number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
#The else block executes only if no exception occurs.
try:
    num = int(input("Enter a number: "))
except ValueError:
    print("Invalid input")
else:
    print("You entered:", num)
```

```python
#The finally block always executes, whether an exception occurs or not.
try:
    file = open("sampe.txt", "r")
    content = file.read()

except FileNotFoundError:
    print("File not found")

finally:
    print("Execution completed")
```

```python
#You can generate exceptions manually using raise.
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
```

```python
#The assert statement is used to check whether a condition is true. If the condition is False, Python raises an AssertionError. Use assert for internal sanity checks and debugging.
age = 20
assert age > 100,"greater than exp"
print("valid age")
```

```python
try:
    age = -1
    assert age >= 0, "Age cannot be negative"

except AssertionError as e:
    print("Assertion failed:", e)
```

assert internally raises an AssertionError when a condition is false. raise is more general and can raise any exception. Assertions are mainly used for debugging and sanity checks, while raise is used for explicit error handling and validation in production code.

```python
#A nested try-except means placing one try-except block inside another try, except, else, or finally block.
try:
    print("Outer try")

    try:
        num = int("abc")
    except ValueError:
        print("Inner exception handled")

except Exception:
    print("Outer exception handled")
```

```python
#A user-defined exception is a custom exception class created by the programmer to represent application-specific errors.
class InvalidAgeError(Exception):
    pass

age = -5

if age < 0:
    raise InvalidAgeError("Age cannot be negative")
```

```python
#Custom Exception with Constructor
class InsufficientBalanceError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(
            f"Balance {balance} is insufficient for withdrawal of {amount}"
        )

balance = 1000
withdraw = 1500

if withdraw > balance:
    raise InsufficientBalanceError(balance, withdraw)
```

```python
#Exception chaining is the process of raising a new exception while preserving information about the original exception that caused it.
#This helps in debugging because you can see the entire chain of errors instead of only the final one.
try:
    num = int("abc")
except ValueError as e:
    raise RuntimeError("Conversion failed") from e

'''ValueError = original exception
RuntimeError = new exception
e = exception being chained'''
```

types of exceptions

```python
#A TypeError occurs when an operation or function is applied to an object of an inappropriate data type.
try:
    result = 10 + "20"
except TypeError as e:
    print("Type Error:", e)

'''TypeError
Wrong data type.
10 + "20"
'''
```

```python
#FileNotFoundError is raised when Python tries to access a file that does not exist at the specified path.
try:
    file = open("data.txt", "r")

except FileNotFoundError as e:
    print("Error:", e)
```

```python
'''ValueError
Correct type, but invalid value.
int("abc")'''
try:
    age = int(input("Enter age: "))
except ValueError:
    print("Please enter a valid number.")
```

```python
'''An IndexError occurs when you try to access an index that is outside the valid range of a sequence such as a list, tuple, or string.
numbers = [10, 20, 30]

print(numbers[5])
'''
try:
    numbers = [10, 20, 30]
    print(numbers[5])

except IndexError as e:
    print("Error:", e)
```

```python
'''A KeyError occurs when you try to access a dictionary key that does not exist.

student = {
    "name": "John",
    "age": 20
}

print(student["city"])
'''

student = {
    "name": "John",
    "age": 20
}

try:
    print(student["city"])

except KeyError as e:
    print("Key not found:", e)
```


Logging is a way to record events, messages, warnings, and errors that occur while a program is running.
Instead of using many print() statements, professional applications use the logging module to keep track of what is happening in the program.
DEBUG < INFO < WARNING < ERROR < CRITICAL

```python
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.INFO
)

logging.info("Application Started")
logging.error("Error occurred")
```

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
```
